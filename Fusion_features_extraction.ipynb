### mfcc_derivations

```python
import numpy as np
from tqdm import tqdm
from spafe.features.lfcc import lfcc
from scipy.io.wavfile import read
from spafe.features.lfcc import lfcc
from spafe.utils.vis import show_features
import spafe
from spafe.features.mfcc import mfcc
from spafe.utils.preprocessing import SlidingWindow
import librosa

def mfcc_features_extractor(file):
    fs, sig = read(file)
    mfccs  = mfcc(sig,
              fs=fs,
              num_ceps =30,
              pre_emph=1,
              pre_emph_coeff=0.97,
              window=SlidingWindow(0.025, 0.005, "hamming"),
              nfilts=80,
              nfft=2048,
              low_freq=0,
              high_freq=fs/2,
                 )
    data_transpose = np.transpose(mfccs)

    delta_mfccs1 = librosa.feature.delta(data_transpose)
    delta2_mfccs1 = librosa.feature.delta(data_transpose, order=2)

    mfccs_scaled_features1 = np.mean(mfccs,axis=0)
    mfccs_scaled_features2 = np.mean(np.transpose(delta_mfccs1),axis=0)
    mfccs_scaled_features3 = np.mean(np.transpose(delta2_mfccs1),axis=0)

    concatenated_arr = np.concatenate((mfccs_scaled_features1, mfccs_scaled_features2, mfccs_scaled_features3))
    
    return concatenated_arr
```

```python
import numpy as np
from tqdm import tqdm
import spafe
from spafe.utils.preprocessing import SlidingWindow
import os
import librosa



audio_dataset_path = '......./Downloads/Tacticas Audio zip/version11012024/Tacticas Audio/20240123_remove_slience/'


extracted_features=[]
for index_num,row in tqdm(frame.iterrows()):
    file_name = os.path.join(os.path.abspath(audio_dataset_path),str(row["name"]))
    final_class_labels=row["label1"]
    name = row["name"]
    excercise = row["excercise"]
    userid = row["userid"]
    day_index = row["date"]
    
    if os.path.exists(file_name):
        audio, sample_rate = librosa.load(file_name , res_type='kaiser_fast') 
        length = librosa.get_duration(y=audio, sr=sample_rate)
        if length > 0.2:
            data = mfcc_features_extractor(file_name)
            extracted_features.append([data,final_class_labels,name,userid,excercise,length,day_index])
```

```python
df_features = pd.DataFrame(extracted_features,columns=['mfcc','label','name','userid','excercise','length','date'])
df_features
```

```python
df_features['new_feature'] = df_features['mfcc'].apply(lambda x: x[:90])
df_features
```

```python
X = df_features['new_feature'].to_list()
y = df_features['label'].to_list()

import numpy as np

X = np.vstack(X)
X.shape
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import precision_score, roc_auc_score, f1_score, recall_score, confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Assume your DataFrame is df with columns 'new_feature' and 'label'
scaler = StandardScaler()


model = SVC()  # Adjust the number of neighbors as needed

# Initialize StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

# Initialize an empty list to store metric scores
precision_scores = []
auc_scores = []
f1_scores = []
sensitivity_scores = []
specificity_scores = []
recall_scores = []
accuracy_scores = []

# Loop through each fold
for train_index, test_index in skf.split(X, y):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = np.array(y)[train_index], np.array(y)[test_index]

    # Scale the features using StandardScaler
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Train the KNN model
    model.fit(X_train_scaled, y_train)

    # Make predictions on the scaled test set
    y_pred = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)
    accuracy_scores.append(accuracy)

    # Calculate precision
    precision = precision_score(y_test, y_pred)
    precision_scores.append(precision)

    # Calculate AUC
    auc = roc_auc_score(y_test, y_pred)
    auc_scores.append(auc)

    # Calculate F1 score
    f1 = f1_score(y_test, y_pred)
    f1_scores.append(f1)

    # Calculate sensitivity (recall)
    sensitivity = recall_score(y_test, y_pred)
    sensitivity_scores.append(sensitivity)

    # Calculate specificity
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
    specificity = tn / (tn + fp)
    specificity_scores.append(specificity)

    # Calculate recall
    recall = recall_score(y_test, y_pred)
    recall_scores.append(recall)

# Calculate standard deviations
sd_precision = np.std(precision_scores)
sd_auc = np.std(auc_scores)
sd_f1 = np.std(f1_scores)
sd_sensitivity = np.std(sensitivity_scores)
sd_specificity = np.std(specificity_scores)
sd_recall = np.std(recall_scores)
sd_accuracy = np.std(accuracy_scores)

# Print the mean and standard deviation scores across all folds

print(f'Accuracy: {100*np.mean(accuracy_scores):0.2f} ({100*sd_accuracy:0.2f})')
print(f'Sensitivity: {100*np.mean(sensitivity_scores):0.2f} ({100*sd_sensitivity:0.2f})')
print(f'Specificity: {100*np.mean(specificity_scores):0.2f} ({100*sd_specificity:0.2f})')
print(f'AUC: {100*np.mean(auc_scores):0.2f} ({100*sd_auc:0.2f})')
print(f'Precision: {100*np.mean(precision_scores):0.2f} ({100*sd_precision:0.2f})')
print(f'F1 Score: {100*np.mean(f1_scores):0.2f} ({100*sd_f1:0.2f})')
```
